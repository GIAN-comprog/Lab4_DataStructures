```python
# APPENDIX: ADDITIONAL EXERCISE 1 - Engineering Test Data Organizer

LAST_NAME = "BERNARDO"
SEED_NUM = 9
FAVORITE_ARTIST = "RADIOHEAD"

categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]
sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
raw_values = [SEED_NUM + (ord(char) % 40) for char in FAVORITE_ARTIST.replace(" ", "")]

test_records = []
val_idx = 0
for i, src in enumerate(sources):
    for j, cat in enumerate(categories):
        val = float(raw_values[val_idx % len(raw_values)] + (i * 2) + j)
        test_records.append({"Source": src, "Category": cat, "Value": val})
        val_idx += 1

# Inject Mandatory Repeated Records using valid element indices
test_records.append({"Source": f"{LAST_NAME}_T1", "Category": "TEMPERATURE", "Value": float(raw_values[0])})
test_records.append({"Source": f"{LAST_NAME}_T1", "Category": "TEMPERATURE", "Value": float(SEED_NUM + 45)})
test_records.append({"Source": f"{LAST_NAME}_T2", "Category": "VOLTAGE", "Value": float(SEED_NUM + 55)})

test_records.sort(key=lambda x: (x["Source"], x["Category"]))

print("=== 1. GENERATED & ORGANIZED TEST DATA ===")
print(f"{'SOURCE':<15} | {'CATEGORY':<15} | {'VALUE':<10}")
print("-" * 48)
for record in test_records:
    print(f"{record['Source']:<15} | {record['Category']:<15} | {record['Value']:<10.2f}")

distinct_sources = set(r["Source"] for r in test_records)
distinct_categories = set(r["Category"] for r in test_records)

print("\n=== 2. REPEATED/DISTINCT DATA ANALYSIS ===")
print(f"Distinct Sources Count: {len(distinct_sources)} {list(distinct_sources)}")
print(f"Distinct Categories Count: {len(distinct_categories)} {list(distinct_categories)}")

seen_combinations = {}
print("\nIdentified Repeated/Multiple Records:")
for r in test_records:
    key = (r["Source"], r["Category"])
    if key in seen_combinations:
        seen_combinations[key].append(r["Value"])
    else:
        seen_combinations[key] = [r["Value"]]

for key, values in seen_combinations.items():
    if len(values) > 1:
        print(f"Source/Category: {key} -> Values: {values}")

print("\n=== 3. FINAL CONSOLIDATED SUMMARY REPORT ===")
print(f"{'CATEGORY':<15} | {'MEAN':<8} | {'MIN':<8} | {'MAX':<8} | {'COUNT':<5}")
print("-" * 52)
for cat in categories:
    cat_values = [r["Value"] for r in test_records if r["Category"] == cat]
    if cat_values:
        print(f"{cat:<15} | {sum(cat_values)/len(cat_values):<8.2f} | {min(cat_values):<8.2f} | {max(cat_values):<8.2f} | {len(cat_values):<5}")    
```

```python
# APPENDIX: ADDITIONAL EXERCISE 2 - Student Performance Data Analyzer

LAST_NAME = "BERNARDO"
SEED_NUM = 9
FAVORITE_ARTIST = "RADIOHEAD"

students = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]
raw_scores = [50 + (SEED_NUM + ord(char)) % 51 for char in FAVORITE_ARTIST.replace(" ", "")]

performance_records = []
score_idx = 0
for i, stu in enumerate(students):
    for j, sub in enumerate(subjects):
        score = raw_scores[score_idx % len(raw_scores)] + (i * 2) - j
        performance_records.append({"Student": stu, "Subject": sub, "Score": float(max(50, min(100, score)))})
        score_idx += 1

performance_records.append({"Student": f"{LAST_NAME}_S1", "Subject": "PROGRAMMING", "Score": float(raw_scores[0])})
performance_records.append({"Student": f"{LAST_NAME}_S5", "Subject": "ENGLISH", "Score": None})

performance_records.sort(key=lambda x: (x["Student"], x["Subject"]))

print("=== 1. GENERATED STUDENT PERFORMANCE DATA ===")
print(f"{'STUDENT':<15} | {'SUBJECT':<15} | {'SCORE':<8} | {'CATEGORY':<12}")
print("-" * 58)

for r in performance_records:
    score = r["Score"]
    if score is None: cat = "INCOMPLETE"
    elif score >= 90: cat = "EXCELLENT"
    elif score >= 75: cat = "GOOD"
    elif score >= 60: cat = "PASS"
    else: cat = "FAIL"
    print(f"{r['Student']:<15} | {r['Subject']:<15} | {f'{score:.2f}' if score is not None else 'None':<8} | {cat:<12}")

print("\n=== 2. REPEATED & INCOMPLETE RECORDS ANALYSIS ===")
seen_pairs = {}
for r in performance_records:
    pair = (r["Student"], r["Subject"])
    seen_pairs[pair] = seen_pairs.get(pair, 0) + 1
    if seen_pairs[pair] > 1:
        print(f"-> Student/Subject: {pair} appeared multiple times.")

for r in performance_records:
    if r["Score"] is None:
        print(f"-> Student: {r['Student']} | Subject: {r['Subject']} is Incomplete.")

print("\n=== 3. STUDENT & SUBJECT PERFORMANCE SUMMARIES ===")
print("\n--- Student Averages ---")
for stu in students:
    scores = [r["Score"] for r in performance_records if r["Student"] == stu and r["Score"] is not None]
    print(f"{stu}: {sum(scores)/len(scores):.2f}" if scores else f"{stu}: 0.00")

print("\n--- Subject Averages ---")
for sub in subjects:
    scores = [r["Score"] for r in performance_records if r["Subject"] == sub and r["Score"] is not None]
    print(f"{sub}: {sum(scores)/len(scores):.2f}" if scores else f"{sub}: 0.00")    
```

```python
# APPENDIX: ADDITIONAL EXERCISE 3 - Transaction History Analyzer

LAST_NAME = "BERNARDO"
SEED_NUM = 9
FAVORITE_ARTIST = "RADIOHEAD"

# Define configurations
product_catalog = {
    "NOTEBOOK": 45.0, 
    "PEN": 15.0, 
    "USB": 350.0, 
    "HEADPHONES": 599.0, 
    "CALCULATOR": 850.0
}
customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
products = list(product_catalog.keys())

# Generate dynamic quantities based on artist characters
raw_quantities = [1 + (SEED_NUM + ord(char)) % 5 for char in FAVORITE_ARTIST.replace(" ", "")]

transaction_records = []
qty_idx = 0

# Assemble records
for i, cust in enumerate(customers):
    for j, prod in enumerate(products):
        qty = raw_quantities[qty_idx % len(raw_quantities)]
        transaction_records.append({
            "Customer": cust, 
            "Product": prod, 
            "Quantity": qty, 
            "Price": product_catalog[prod], 
            "Value": qty * product_catalog[prod]
        })
        qty_idx += 1

# Fulfill requirement: Include one repeated transaction record
dup_qty = int(raw_quantities[0])
transaction_records.append({
    "Customer": f"{LAST_NAME}_C1", 
    "Product": "NOTEBOOK", 
    "Quantity": dup_qty, 
    "Price": product_catalog["NOTEBOOK"], 
    "Value": float(dup_qty * product_catalog["NOTEBOOK"])
})

# Sort items cleanly
transaction_records.sort(key=lambda x: (x["Customer"], x["Product"]))

# --- OUTPUT PRINTS ---
print("=== 1. GENERATED TRANSACTION RECORDS ===")
print(f"{'CUSTOMER':<15} | {'PRODUCT':<12} | {'QTY':<4} | {'PRICE':<8} | {'VALUE':<10}")
print("-" * 59)
for r in transaction_records:
    print(f"{r['Customer']:<15} | {r['Product']:<12} | {r['Quantity']:<4} | {r['Price']:<8.2f} | {r['Value']:<10.2f}")

print("\n=== 2. REPEATED/DISTINCT TRANSACTIONS ANALYSIS ===")
seen_tx = {}
for r in transaction_records:
    pair = (r["Customer"], r["Product"])
    seen_tx[pair] = seen_tx.get(pair, 0) + 1
    if seen_tx[pair] > 1:
        print(f"-> {r['Customer']} purchased {r['Product']} multiple times.")

print("\n=== 3. CUSTOMER & PRODUCT SUMMARIES ===")
for cust in customers:
    total_cust = sum(r['Value'] for r in transaction_records if r['Customer'] == cust)
    print(f"{cust} Total Spent: Php {total_cust:,.2f}")
print("")

for prod in products:
    tot_qty = sum(r['Quantity'] for r in transaction_records if r['Product'] == prod)
    tot_rev = sum(r['Value'] for r in transaction_records if r['Product'] == prod)
    print(f"{prod:<12} -> Total Qty: {tot_qty:<4} | Revenue: Php {tot_rev:,.2f}")

overall_total = sum(r['Value'] for r in transaction_records)
print(f"\nOVERALL DATASET TOTAL TRANSACTION VALUE: Php {overall_total:,.2f}")
```
